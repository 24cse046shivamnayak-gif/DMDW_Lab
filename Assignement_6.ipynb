```python
import pandas as pd
import numpy as np
```

```python
# for data preprocessing
from sklearn import preprocessing
```

```python
#create a dataset
data= pd.DataFrame({
    'age':[15,16,17,18,19,20],
    'study time': [1,2,3,2,4,3],
    'failure': [0,1,0,2,1,0]
}).astype(float)
```

```python
#min=max normalization
#converts the values into the range [o,1]
minmax_scaler = preprocessing.MinMaxScaler()
minmax_scaled= minmax_scaler.fit_transform(data)
```

```python
#Z-score Normalization
#Mean= 0 and Standard Devaition =1
zscore_scaled = preprocessing.scale(data)
```

```python
#Decimal Scaling Normalization
j= np.ceil(np.log10(data.abs().max()))
decimal_scaled= data/(10**j)
```

```python
#dispaly original data
print("====ORIGINAL DATA====")
print(data)
```

```python
#Display MinMax Normalization
print("====MINMAX NORMAILIZATION====")
print(pd.DataFrame(minmax_scaled,columns=data.columns))
```

```python
#Display Z-Score Normalization
print("====Z-SCORE NORMALIZATION====")
print(pd.DataFrame(zscore_scaled,columns=data.columns))
```

```python
#Display Decimal Scaling
print("====DECIAML SCALING NORMALIZATION====")
print(decimal_scaled)
```
