```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sbs
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
df=pd.read_csv("/content/student-mat.csv",sep=";")
df.head()
df.info()
```

```python
dfn = df[['traveltime', 'studytime']]
x = dfn['traveltime']
y = dfn['studytime']
sbs.lineplot(x=x, y=y, dashes=True)
plt.show()
```

```python
#CORRELATION ANALYSIS
correlation=dfn.corr()
print(correlation)
sbs.heatmap(correlation,cmap="YlGnBu")
plt.show()
```

```python
#COVARIANCE ANALYSIS
covar=dfn.cov()
print(covar)
sbs.heatmap(covar,cmap="YlGnBu")
plt.show()
```
